```python
from datasets import load_dataset

dataset = load_dataset("mteb/banking77")

print(dataset)
print(dataset["train"][0])
```

```python
dataset
```

```python
label_names = sorted(
    set(dataset["train"]["label_text"])
)

print(len(label_names))
print(label_names[:20])
```

```python
chosen_labels = [
    "activate_my_card",
    "age_limit",
    "atm_support",
    "cancel_transfer",
    "card_arrival",
    "card_not_working",
    "cash_withdrawal_charge",
    "change_pin",
    "compromised_card",
    "cash_withdrawal_not_recognised",
]

train_ds = dataset["train"].filter(
    lambda x: x["label_text"] in chosen_labels
)

test_ds = dataset["test"].filter(
    lambda x: x["label_text"] in chosen_labels
)

print(len(train_ds))
print(len(test_ds))
print(train_ds[0])
```
